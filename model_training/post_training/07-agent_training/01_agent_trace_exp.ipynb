# Agent Trace Dataset Exploration

Inspect the dataset structure, unique values, and the messages and evaluation result for a recorded agent attempt.

```python
from datasets import load_dataset
```

## 1. Load the dataset

Load the `train.raw` split and select the first 10 records for sample inspection. The unique-value cells below use the full loaded split.

```python
dataset = load_dataset('SWE-Gym/OpenHands-Sampled-Trajectories', split="train.raw")
samples = list(dataset.take(10) )
```

## 3. Explore unique values

Inspect issue IDs, run IDs, and outcome values across the loaded split.

```python
unique_instance_ids = sorted({
    row["instance_id"]
    for row in dataset
    if row.get("instance_id")
})

print(f"{len(unique_instance_ids)} unique instance IDs")
unique_instance_ids
```

### Run IDs

The printed label in this existing cell says instance IDs; its values are run IDs.

```python
run_instance_ids = sorted({
    row["run_id"]
    for row in dataset
    if row.get("run_id")
})

print(f"{len(run_instance_ids)} unique instance IDs")
run_instance_ids
```

### Resolved values

The existing condition filters out `False`, so this cell does **not** show all outcome values or their distribution.

```python
resolved_instance_ids = sorted({
    row["resolved"]
    for row in dataset
    if row.get("resolved")
})

print(f"{len(resolved_instance_ids)} unique instance IDs")
resolved_instance_ids
```

## 5. Inspect tools and messages

The tool definitions cell is currently commented out. The trace display below prints the first sample’s messages and tool calls in order; `trace_1` contains the second sample’s messages.

```python
# samples[0]['tools']
```

```python
trace_1 = samples[1]["messages"]
```

```python
import json

trace = samples[0]["messages"]

for i, message in enumerate(trace):
    print(f"\n--- Step {i} | {message.get('role', 'unknown')} ---")

    content = message.get("content")
    if content:
        print(content)

    for call in message.get("tool_calls") or []:
        function = call.get("function", {})
        print(f"Tool: {function.get('name')}")
        print("Arguments:")
        print(json.dumps(function.get("arguments"), indent=2))
```
