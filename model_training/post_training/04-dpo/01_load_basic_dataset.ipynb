### Step-1:- Load a Dataset

```python
DATASET_ID = "trl-lib/ultrafeedback_binarized"

train_data = load_dataset(DATASET_ID, split="train")
test_data = load_dataset(DATASET_ID, split="test")

print(train_data)
print(train_data.column_names)
```

```python
samples= list(train_data.select(range(5)))
example = samples[0]
assert example["chosen"][:-1] == example["rejected"][:-1]
```

```python
prompt = example["chosen"][:-1]
chosen = example["chosen"][-1]["content"]
rejected = example["rejected"][-1]["content"]

print("PROMPT:", type(prompt), prompt)
print("\nCHOSEN:", chosen[:500])
print("\nREJECTED:", rejected[:500])

print("\nScores:",
      example["score_chosen"],
      example["score_rejected"])
```
