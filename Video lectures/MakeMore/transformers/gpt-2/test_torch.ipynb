```python
import torch
import torch.nn as nn
import tiktoken 
```

```python
x = "Hi, My name is Vedaang Chopra"
```

```python
tokenizer = tiktoken.get_encoding("gpt2")
enc_x = tokenizer.encode(x)
enc_x = torch.tensor(enc_x)
enc_x = enc_x.unsqueeze(0).repeat(5, 1).float()
enc_x.shape, enc_x.dtype
```

```python
n_embd = 10
```

```python
c_attn = nn.Linear(n_embd, 3 * n_embd)
c_attn(enc_x).shape
```
