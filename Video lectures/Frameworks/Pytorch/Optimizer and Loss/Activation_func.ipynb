## Tests and Assertions

Run these checks after you finish the TODOs above.

The tests below verify:
- output values
- derivative values
- output shapes
- consistency with either the exact GELU or a common approximation


```python
def assert_numpy_close(actual, expected, *, name, atol=1e-6, rtol=1e-6):
    assert isinstance(actual, np.ndarray), f"{name} should return a NumPy array"
    assert actual.shape == expected.shape, (
        f"{name} returned shape {actual.shape}, expected {expected.shape}"
    )
    assert np.allclose(actual, expected, atol=atol, rtol=rtol), (
        f"{name} values do not match expected results"
    )


def assert_torch_close(actual, expected, *, name, atol=1e-6, rtol=1e-6):
    assert isinstance(actual, torch.Tensor), f"{name} should return a torch.Tensor"
    assert actual.shape == expected.shape, (
        f"{name} returned shape {tuple(actual.shape)}, expected {tuple(expected.shape)}"
    )
    assert torch.allclose(actual, expected, atol=atol, rtol=rtol), (
        f"{name} values do not match expected results"
    )


def assert_matches_either(actual, option_a, option_b, *, name, atol=1e-4, rtol=1e-4):
    matches_a = np.allclose(actual, option_a, atol=atol, rtol=rtol)
    matches_b = np.allclose(actual, option_b, atol=atol, rtol=rtol)
    assert matches_a or matches_b, (
        f"{name} did not match the exact GELU or the tanh approximation"
    )
```

```python
def run_numpy_activation_tests():
    x = np.array([-3.0, -1.0, -0.5, 0.0, 0.5, 1.0, 3.0], dtype=np.float64)
    x_no_zero = np.array([-3.0, -1.0, -0.5, 0.5, 1.0, 3.0], dtype=np.float64)
    alpha_leaky = 0.1
    alpha_elu = 1.0

    sigmoid_expected = 1.0 / (1.0 + np.exp(-x))
    sigmoid_derivative_expected = sigmoid_expected * (1.0 - sigmoid_expected)
    assert_numpy_close(sigmoid_numpy(x), sigmoid_expected, name="sigmoid_numpy")
    assert_numpy_close(
        sigmoid_derivative_numpy(x),
        sigmoid_derivative_expected,
        name="sigmoid_derivative_numpy",
    )

    tanh_expected = np.tanh(x)
    tanh_derivative_expected = 1.0 - tanh_expected ** 2
    assert_numpy_close(tanh_numpy(x), tanh_expected, name="tanh_numpy")
    assert_numpy_close(
        tanh_derivative_numpy(x),
        tanh_derivative_expected,
        name="tanh_derivative_numpy",
    )

    relu_expected = np.maximum(0.0, x)
    relu_derivative_expected = np.where(x_no_zero > 0.0, 1.0, 0.0)
    assert_numpy_close(relu_numpy(x), relu_expected, name="relu_numpy")
    assert_numpy_close(
        relu_derivative_numpy(x_no_zero),
        relu_derivative_expected,
        name="relu_derivative_numpy",
    )

    leaky_relu_expected = np.where(x > 0.0, x, alpha_leaky * x)
    leaky_relu_derivative_expected = np.where(x_no_zero > 0.0, 1.0, alpha_leaky)
    assert_numpy_close(
        leaky_relu_numpy(x, alpha=alpha_leaky),
        leaky_relu_expected,
        name="leaky_relu_numpy",
    )
    assert_numpy_close(
        leaky_relu_derivative_numpy(x_no_zero, alpha=alpha_leaky),
        leaky_relu_derivative_expected,
        name="leaky_relu_derivative_numpy",
    )

    elu_expected = np.where(x > 0.0, x, alpha_elu * (np.exp(x) - 1.0))
    elu_derivative_expected = np.where(x_no_zero > 0.0, 1.0, alpha_elu * np.exp(x_no_zero))
    assert_numpy_close(elu_numpy(x, alpha=alpha_elu), elu_expected, name="elu_numpy")
    assert_numpy_close(
        elu_derivative_numpy(x_no_zero, alpha=alpha_elu),
        elu_derivative_expected,
        name="elu_derivative_numpy",
    )

    x_torch = torch.tensor(x, dtype=torch.float64)
    gelu_exact = torch.nn.functional.gelu(x_torch, approximate='none').numpy()
    gelu_approx = torch.nn.functional.gelu(x_torch, approximate='tanh').numpy()
    gelu_actual = gelu_numpy(x)
    assert isinstance(gelu_actual, np.ndarray), "gelu_numpy should return a NumPy array"
    assert gelu_actual.shape == x.shape, "gelu_numpy should preserve input shape"
    assert_matches_either(gelu_actual, gelu_exact, gelu_approx, name="gelu_numpy")

    eps = 1e-5
    gelu_numeric_derivative = (gelu_numpy(x + eps) - gelu_numpy(x - eps)) / (2.0 * eps)
    assert_numpy_close(
        gelu_derivative_numpy(x),
        gelu_numeric_derivative,
        name="gelu_derivative_numpy",
        atol=1e-4,
        rtol=1e-4,
    )

    print("All NumPy activation tests passed.")
```

```python
def run_torch_activation_tests():
    x = torch.tensor([-3.0, -1.0, -0.5, 0.0, 0.5, 1.0, 3.0], dtype=torch.float64)
    x_no_zero = torch.tensor([-3.0, -1.0, -0.5, 0.5, 1.0, 3.0], dtype=torch.float64)
    alpha_leaky = 0.1
    alpha_elu = 1.0

    sigmoid_expected = 1.0 / (1.0 + torch.exp(-x))
    sigmoid_derivative_expected = sigmoid_expected * (1.0 - sigmoid_expected)
    assert_torch_close(sigmoid_torch(x), sigmoid_expected, name="sigmoid_torch")
    assert_torch_close(
        sigmoid_derivative_torch(x),
        sigmoid_derivative_expected,
        name="sigmoid_derivative_torch",
    )

    tanh_expected = torch.tanh(x)
    tanh_derivative_expected = 1.0 - tanh_expected ** 2
    assert_torch_close(tanh_torch(x), tanh_expected, name="tanh_torch")
    assert_torch_close(
        tanh_derivative_torch(x),
        tanh_derivative_expected,
        name="tanh_derivative_torch",
    )

    relu_expected = torch.maximum(torch.zeros_like(x), x)
    relu_derivative_expected = torch.where(x_no_zero > 0.0, torch.ones_like(x_no_zero), torch.zeros_like(x_no_zero))
    assert_torch_close(relu_torch(x), relu_expected, name="relu_torch")
    assert_torch_close(
        relu_derivative_torch(x_no_zero),
        relu_derivative_expected,
        name="relu_derivative_torch",
    )

    leaky_relu_expected = torch.where(x > 0.0, x, alpha_leaky * x)
    leaky_relu_derivative_expected = torch.where(
        x_no_zero > 0.0,
        torch.ones_like(x_no_zero),
        torch.full_like(x_no_zero, alpha_leaky),
    )
    assert_torch_close(
        leaky_relu_torch(x, alpha=alpha_leaky),
        leaky_relu_expected,
        name="leaky_relu_torch",
    )
    assert_torch_close(
        leaky_relu_derivative_torch(x_no_zero, alpha=alpha_leaky),
        leaky_relu_derivative_expected,
        name="leaky_relu_derivative_torch",
    )

    elu_expected = torch.where(x > 0.0, x, alpha_elu * (torch.exp(x) - 1.0))
    elu_derivative_expected = torch.where(
        x_no_zero > 0.0,
        torch.ones_like(x_no_zero),
        alpha_elu * torch.exp(x_no_zero),
    )
    assert_torch_close(elu_torch(x, alpha=alpha_elu), elu_expected, name="elu_torch")
    assert_torch_close(
        elu_derivative_torch(x_no_zero, alpha=alpha_elu),
        elu_derivative_expected,
        name="elu_derivative_torch",
    )

    gelu_exact = torch.nn.functional.gelu(x, approximate='none')
    gelu_approx = torch.nn.functional.gelu(x, approximate='tanh')
    gelu_actual = gelu_torch(x)
    assert isinstance(gelu_actual, torch.Tensor), "gelu_torch should return a torch.Tensor"
    assert gelu_actual.shape == x.shape, "gelu_torch should preserve input shape"
    gelu_matches_exact = torch.allclose(gelu_actual, gelu_exact, atol=1e-4, rtol=1e-4)
    gelu_matches_approx = torch.allclose(gelu_actual, gelu_approx, atol=1e-4, rtol=1e-4)
    assert gelu_matches_exact or gelu_matches_approx, (
        "gelu_torch did not match the exact GELU or the tanh approximation"
    )

    eps = 1e-5
    gelu_numeric_derivative = (gelu_torch(x + eps) - gelu_torch(x - eps)) / (2.0 * eps)
    assert_torch_close(
        gelu_derivative_torch(x),
        gelu_numeric_derivative,
        name="gelu_derivative_torch",
        atol=1e-4,
        rtol=1e-4,
    )

    print("All PyTorch activation tests passed.")
```
